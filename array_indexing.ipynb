```python
# Numpy Indexing and Slicing of Vectors
import numpy as np
```

```python
arr = np.arange(1, 21)
arr
```

```python
arr[6]
```

```python
arr[1:5]
```

```python
arr[:5]
```

```python
arr[:]
```

```python
arr[3:]
```

```python
arr_slice = arr[3::2]
arr_slice
```

```python
# Numpy Indexing and Slicing of Matrix
```

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0]
```

```python
arr[5]
```

```python
arr[0,0]
```

```python
arr[0,4]
```

```python
arr[5,4]
```

```python
arr[0:2,1:3]
```

```python
arr_slice = arr[3:,3:]
arr_slice
```

```python
arr[:,2]
```
